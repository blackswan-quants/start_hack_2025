```python
# Relplot: Temperatura Media vs Precipitazioni con Hue=Land Cover
sns.relplot(x='Temp_media', y='Precipitazioni', hue='Land_Cover', data=df)
plt.title("Temp Media vs Precipitazioni (Hue: Land Cover)")
plt.show()
```

```python
# Relplot: Pressione vs Umidità con Hue=Regione
sns.relplot(x='Pressione', y='Umidità', hue='ADM2_1', data=df)
plt.title("Pressione vs Umidità (Hue: Regione)")
plt.show()
```

```python
# Relplot: Vento_2m vs Umidità con Hue=Land Cover
sns.relplot(x='Vento_2m', y='Umidità', hue='Land_Cover', data=df)
plt.title("Vento_2m vs Umidità (Hue: Land Cover)")
plt.show()
# inutile in culo
```

```python
# Relplot: Umidità vs Precipitazioni con Hue=Land Cover (per siccità)
sns.relplot(x='Umidità', y='Precipitazioni', hue='Land_Cover', data=df)
plt.title("Umidità vs Precipitazioni (Hue: Land Cover) - Siccità")
plt.show()
# we have places with a lot of water that are barren, we need to raget them
```

```python
# Relplot: Radiazione Solare vs Temp Superficie con Hue=Land Cover (per land degradation)
sns.relplot(x='Radiazione_Solare', y='Temp_superficie', hue='Land_Cover', data=df)
plt.title("Radiazione Solare vs Temp Superficie (Hue: Land Cover) - Land Degradation")
plt.show()
# on average barren lands have lower temp_superficie
```

```python
sns.relplot(x='distance_to_whole_line', y='distance_to_water', hue='Land_Cover', data=df)
plt.title("Distanza stade vs Distamza acqua (Hue: land cover) - Isolamento e desertificazione")
plt.show()
# on average barren land is further to water than greenland, in general the
# further from the roads the greener the land is (civilisation ruins land), also this becomes irrelevant when near the water.
```

```python
# Relplot: Vento_2m vs Vento_50m con Hue=Land Cover
sns.relplot(x='Vento_2m', y='Umidità', hue='Land_Cover', data=df)
plt.title("Vento_2m vs Vento_50m (Hue: Land Cover)")
plt.show()
```

```python
# Barplot: Land Cover vs Precipitazioni Medie
sns.barplot(x='Land_Cover', y='Precipitazioni', data=df, estimator=np.mean)
plt.title("Precipitazioni Medie per Land Cover")
plt.show()
# differene in water doesn't mean a miningfull differene in Land Cover
```

```python
# Barplot: Regione vs Umidità Media
sns.barplot(x='ADM2_1', y='Umidità', data=df, estimator=np.mean)
plt.title("Umidità Media per Regione")
plt.show()
```

```python
# Barplot: Distretto vs Temperatura Massima
sns.barplot(x='ADM2_1', y='Temp_max', data=df, estimator=np.mean)
plt.xticks(rotation=90)
plt.title("Temperatura Massima per Distretto")
plt.show()
```

```python
#Seleziona le colonne specifiche per il box plot
selected_cols = ['Precipitazioni', 'Temp_media', 'Umidità']

# 3. Box Plot per le variabili selezionate
plt.figure(figsize=(12, 8))
sns.boxplot(data=df[selected_cols])
plt.title('Box Plot di Precipitazioni, Temp_media, Umidità')
plt.xticks(rotation=90)
plt.show()

# there are places where it rains a lot and other where it doesn't an issue to tackle is water distribution and retention in arid areas.
```

```python


# Gestione dei missing values: imputazione con la mediana per tutte le colonne numeriche
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# Normalizzazione delle variabili numeriche
scaler = StandardScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])



# 5. Pair Plot per tutte le variabili numeriche
sns.pairplot(df[numeric_cols])
plt.suptitle('Pair Plot delle variabili numeriche', y=1.02)
plt.show()
```
